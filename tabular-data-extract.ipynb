```python
import pandas as pd
```

```python
df = pd.read_csv("hf://datasets/tolu07/Mental_Health_FAQ/Mental_Health_FAQ.csv")
```

```python
df.head(5)
```

```python
import json
from transformers import pipeline

classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli", device='cuda')
candidate_labels = [
    "Depression",
    "Anxiety",
    "Therapy & Treatment",
    "Medication",
    "Symptoms & Diagnosis",
    "Crisis Support",
    "Eating Disorders",
    "Substance Abuse",
    "Self-Care & Coping Strategies",
    "Mental Health in Children & Adolescents",
    "Mental Health in Older Adults",
    "Workplace Mental Health",
    "Stigma & Awareness",
    "Suicide Prevention",
    "Post-Traumatic Stress Disorder (PTSD)",
    "Bipolar Disorder",
    "Schizophrenia",
    "Obsessive-Compulsive Disorder (OCD)",
    "Attention-Deficit/Hyperactivity Disorder (ADHD)",
    "Borderline Personality Disorder (BPD)",
    "Autism Spectrum Disorder (ASD)",
    "Grief & Loss",
    "Mindfulness & Meditation",
    "Sleep Disorders",
    "Self-Harm",
    "Body Image & Self-Esteem",
    "Trauma & Abuse",
    "General Mental Health Awareness",
]


def assign_topic_ai(question):
    result = classifier(question, candidate_labels)
    return result["labels"][0]
```

```python
json_data = []

for _, row in df.iterrows():
    entry = {
        "id": str(row["Question_ID"]),
        "topic": assign_topic_ai(row["Questions"]),
        "question": str(row["Questions"]),
        "answer": str(row["Answers"]),
        "source": "HuggingFace",
        "is_synthetic": False,
        "url": "tolu07/Mental_Health_FAQ",
    }
    json_data.append(entry)

# Save to output JSON file
with open("Mental_Health_FAQ.json", "w", encoding="utf-8") as f:
    json.dump(json_data, f, indent=4, ensure_ascii=False)

print("Conversion complete! File saved as Mental_Health_FAQ.json")
```
